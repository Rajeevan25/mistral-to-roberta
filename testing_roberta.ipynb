```python
import pandas as pd
from transformers import AutoTokenizer, AutoModelForSequenceClassification, pipeline
from sklearn.metrics import accuracy_score
```

```python
# Load your fine-tuned model and tokenizer
model_name = "RajeevanL/Distill_Roberta"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)
```

```python
# Load test dataset
df = pd.read_csv("/content/twittersentiment_1_1000.csv")
texts = df["text"].astype(str).tolist()
true_labels = df['label_text'].tolist()
```

```python
true_labels
```

```python
# Create sentiment analysis pipeline
classifier = pipeline("sentiment-analysis", model="RajeevanL/Distill_Roberta")

# Get predictions
predictions = classifier(texts)
```

```python
#Map model outputs to label names
predicted_labels = [pred['label'] for pred in predictions]
```

```python
predicted_labels
```

```python
# Calculate accuracy
accuracy = accuracy_score(true_labels, predicted_labels)
print(f"Accuracy of Distill_Roberta: {accuracy:.4f}")
```

```python
# Create sentiment analysis pipeline
student_classifier = pipeline("sentiment-analysis", model="FacebookAI/roberta-base")

# Get predictions
predictions = student_classifier(texts)

label_map = {
    "LABEL_0": "negative",
    "LABEL_1": "neutral",
    "LABEL_2": "positive"
}

predicted_labels_student = [label_map[pred['label']] for pred in predictions]

predicted_labels_student
```

```python
accuracy = accuracy_score(true_labels, predicted_labels_student)
print(f"Accuracy of Distill_Roberta: {accuracy:.4f}")
```

```python
# Create sentiment analysis pipeline
teacher_classifier = pipeline("sentiment-analysis", model="mistralai/Mistral-7B-Instruct-v0.1")

# Get predictions
predictions = teacher_classifier(texts)

```
